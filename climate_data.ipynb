```python
import pandas as pd

df = pd.read_csv("climate_finance_data.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.groupby("region")["estimated_cost_million"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

region_cost = df.groupby("region")["estimated_cost_million"].sum()

region_cost.plot(kind="bar")
plt.title("Climate Financial Impact by Region")
plt.xlabel("Region")
plt.ylabel("Total Cost (Million)")
plt.show()
```

```python
df.groupby("sector")["estimated_cost_million"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

sector_cost = df.groupby("sector")["estimated_cost_million"].sum()

sector_cost.plot(kind="bar")
plt.title("Climate Financial Impact by Sector")
plt.xlabel("Sector")
plt.ylabel("Total Cost (Million)")
plt.show()
```

```python
df.groupby("event_type")["estimated_cost_million"].mean().sort_values(ascending=False)
```

```python
event_cost = df.groupby("event_type")["estimated_cost_million"].mean()

event_cost.plot(kind="bar")
plt.title("Average Cost per Climate Event Type")
plt.xlabel("Event Type")
plt.ylabel("Average Cost (Million)")
plt.show()
```
